## Exercise 1

** Follow along with these steps: **
* ** Create a figure object called fig using plt.figure() **
* ** Use add_axes to add an axis to the figure canvas at [0,0,1,1]. Call this new axis ax. **
* ** Plot (x,y) on that axes and set the labels and titles to match the plot below:**

```python
import matplotlib.pyplot as plt

# Données (à définir si ce n’est pas déjà fait dans ton TP)
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]

# 1. Créer la figure
fig = plt.figure()

# 2. Ajouter un axe qui occupe toute la figure
ax = fig.add_axes([0, 0, 1, 1])

# 3. Tracer les données
ax.plot(x, y)

# 4. Ajouter titres et labels
ax.set_xlabel('X')
ax.set_ylabel('Y')
ax.set_title('Titre du graphique')

# Afficher (important si hors Jupyter !)
plt.show()
```

## Exercise 3

** Create the plot below by adding two axes to a figure object at [0,0,1,1] and [0.2,0.5,.4,.4]**

```python
# Données (si pas déjà définies)
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]
z = [0, 1, 8, 27, 64]

fig = plt.figure()

# Grand axe : affiche (x, z)
ax1 = fig.add_axes([0, 0, 1, 1])
ax1.plot(x, z, 'r')  # rouge
ax1.set_xlabel('X')
ax1.set_ylabel('Z')
ax1.set_title('Grand graphique')

# Petit axe : affiche (x, y) avec limites précises
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])  # plus grand que dans ex2
ax2.plot(x, y, 'b')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Petit graphique')
ax2.set_xlim(2, 4)   # zoom sur x de 2 à 4
ax2.set_ylim(4, 16)  # zoom sur y de 4 à 16

plt.show()
```

** Now use x,y, and z arrays to recreate the plot below. Notice the xlimits and y limits on the inserted plot:**

```python
import matplotlib.pyplot as plt

# Données
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]
z = [0, 1, 8, 27, 64]

# Créer la figure
fig = plt.figure()

# Grand axe : [gauche, bas, largeur, hauteur] = [0, 0, 1, 1]
ax1 = fig.add_axes([0, 0, 1, 1])
ax1.plot(x, z, 'r')          # Tracer (x, z) en rouge
ax1.set_xlabel('X')
ax1.set_ylabel('Z')
ax1.set_title('Courbe principale (z = x³)')

# Petit axe inséré : [0.2, 0.5, 0.4, 0.4]
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])
ax2.plot(x, y, 'b')          # Tracer (x, y) en bleu

# Appliquer les limites comme demandé (zoom typique)
ax2.set_xlim(2, 4)
ax2.set_ylim(4, 16)
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Zoom (y = x²)')

# Afficher la figure
fig
```

## Exercise 4

** Use plt.subplots(nrows=1, ncols=2) to create the plot below.**

```python
# Créer 1 ligne, 2 colonnes, avec taille personnalisée
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 4))

# Graphique de gauche : (x, y)
axes[0].plot(x, y, 'b--', linewidth=2)  # trait bleu pointillé
axes[0].set_xlabel('X')
axes[0].set_ylabel('Y')
axes[0].set_title('y = x²')

# Graphique de droite : (x, z)
axes[1].plot(x, z, 'r-', linewidth=3)   # trait rouge plein
axes[1].set_xlabel('X')
axes[1].set_ylabel('Z')
axes[1].set_title('z = x³')

# Ajuste automatiquement l’espacement
plt.tight_layout()

plt.show()
```

** Now plot (x,y) and (x,z) on the axes. Play around with the linewidth and style**

```python
import matplotlib.pyplot as plt

# Données
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]
z = [0, 1, 8, 27, 64]

# Créer 2 sous-graphiques côte à côte, avec une taille personnalisée
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# Premier graphique : (x, y)
axes[0].plot(x, y, color='blue', linestyle='--', linewidth=2.5, marker='o')
axes[0].set_title('y = x²')
axes[0].set_xlabel('X')
axes[0].set_ylabel('Y')

# Deuxième graphique : (x, z)
axes[1].plot(x, z, color='red', linestyle='-', linewidth=3, marker='s')
axes[1].set_title('z = x³')
axes[1].set_xlabel('X')
axes[1].set_ylabel('Z')

# Ajuste l'espacement entre les graphiques
plt.tight_layout()

# Afficher la figure (ou simplement fig dans Jupyter)
fig
```

** See if you can resize the plot by adding the figsize() argument in plt.subplots() are copying and pasting your previous code.**

```python
import matplotlib.pyplot as plt

# Données
x = [0, 1, 2, 3, 4]
y = [0, 1, 4, 9, 16]
z = [0, 1, 8, 27, 64]

# Créer les sous-graphiques AVEC redimensionnement grâce à figsize
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(14, 5))  # ← ici, on a ajouté figsize !

# Premier graphique : (x, y)
axes[0].plot(x, y, color='blue', linestyle='--', linewidth=2.5, marker='o')
axes[0].set_title('y = x²')
axes[0].set_xlabel('X')
axes[0].set_ylabel('Y')

# Deuxième graphique : (x, z)
axes[1].plot(x, z, color='red', linestyle='-', linewidth=3, marker='s')
axes[1].set_title('z = x³')
axes[1].set_xlabel('X')
axes[1].set_ylabel('Z')

# Ajuster l'espacement
plt.tight_layout()

# Afficher la figure
fig
```
