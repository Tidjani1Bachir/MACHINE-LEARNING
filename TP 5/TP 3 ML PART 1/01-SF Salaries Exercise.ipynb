** Import pandas as pd.**

**Lire le fichier Salaries.csv en tant que DataFrame nommé `sal`.**


```python
# Lecture du fichier Salaries.csv dans un DataFrame appelé 'sal'
sal = pd.read_csv('Salaries.csv')
```

**Quelle est la moyenne de la colonne `BasePay` ?**


```python
# Convertir la colonne BasePay en numérique (les valeurs non numériques deviennent NaN)
sal['BasePay'] = pd.to_numeric(sal['BasePay'], errors='coerce')

# Calculer la moyenne de BasePay (ignore les NaN automatiquement)
base_pay_mean = sal['BasePay'].mean()
print("Moyenne de BasePay :", base_pay_mean)
```

**Quel est le montant le plus élevé de `OvertimePay` dans le jeu de données ?**


```python
# Convertir OvertimePay en numérique
sal['OvertimePay'] = pd.to_numeric(sal['OvertimePay'], errors='coerce')

# Trouver la valeur maximale
max_overtime = sal['OvertimePay'].max()
print("Montant le plus élevé de OvertimePay :", max_overtime)
```

**Quel est le titre du poste de JOSEPH DRISCOLL ?**
*Remarque : utilisez des majuscules, sinon vous risquez d’obtenir une réponse incorrecte (il existe également un Joseph Driscoll en minuscules).*


```python
# Filtrer les lignes où le nom est "JOSEPH DRISCOLL"
joseph_row = sal.loc[sal['EmployeeName'] == 'JOSEPH DRISCOLL']

# Afficher le titre du poste (JobTitle)
job_title_joseph = joseph_row['JobTitle'].iloc[0]  # .iloc[0] prend la première valeur
print("Titre du poste de JOSEPH DRISCOLL :", job_title_joseph)
```

**Combien gagne JOSEPH DRISCOLL (y compris les avantages) ?**


```python
# Créer une nouvelle colonne 'TotalPayBenefits' qui est la somme de toutes les colonnes de salaire
sal['TotalPayBenefits'] = sal['BasePay'] + sal['OvertimePay'] + sal['OtherPay'] + sal['Benefits']

# Trouver le TotalPayBenefits de JOSEPH DRISCOLL
joseph_total = sal.loc[sal['EmployeeName'] == 'JOSEPH DRISCOLL', 'TotalPayBenefits'].iloc[0]
print("Salaire total de JOSEPH DRISCOLL (avec avantages) :", joseph_total)
```

**Quel est le nom de la personne la moins payée (y compris les avantages) ?**
*Remarquez-vous quelque chose d’étrange à propos de son salaire ?*


```python
# Trouver l'index de la personne avec le salaire total le plus bas
min_index = sal['TotalPayBenefits'].idxmin()

# Récupérer le nom de cette personne
lowest_paid_person = sal.loc[min_index, 'EmployeeName']
print("Personne la moins payée :", lowest_paid_person)

# Bonus : afficher toute la ligne pour voir pourquoi son salaire est si bas
print("\nDétails de cette personne :")
print(sal.loc[min_index])
```

**Quelle était la moyenne (`mean`) du `BasePay` de tous les employés par année ? (2011-2014)**


```python
# Regrouper par année et calculer la moyenne de BasePay
avg_basepay_by_year = sal.groupby('Year')['BasePay'].mean()
print("Moyenne de BasePay par année :\n", avg_basepay_by_year)
```

**Combien y a-t-il de titres de poste uniques ?**


```python
# Compter le nombre de titres de poste uniques
unique_job_titles = sal['JobTitle'].nunique()
print("Nombre de titres de poste uniques :", unique_job_titles)
```

**Quels sont les 5 titres de poste les plus courants ?**


```python
# Compter les occurrences de chaque JobTitle, trier par ordre décroissant, prendre les 5 premiers
top_5_jobs = sal['JobTitle'].value_counts().head(5)
print("5 titres de poste les plus courants :\n", top_5_jobs)
```

**Combien de titres de poste étaient représentés par une seule personne en 2013 ?**
*(Par exemple, les titres de poste n’apparaissant qu’une seule fois en 2013.)*


```python
# Filtrer les données pour l'année 2013
sal_2013 = sal[sal['Year'] == 2013]

# Compter combien de fois chaque JobTitle apparaît en 2013
job_counts_2013 = sal_2013['JobTitle'].value_counts()

# Compter combien de titres apparaissent exactement une fois
single_person_jobs_2013 = (job_counts_2013 == 1).sum()
print("Nombre de titres de poste représentés par une seule personne en 2013 :", single_person_jobs_2013)
```

**Combien de personnes ont le mot "Chief" dans leur titre de poste ?**
*(C’est assez délicat.)*


```python
# Convertir les colonnes numériques (important pour le filtrage)
sal['BasePay'] = pd.to_numeric(sal['BasePay'], errors='coerce')
sal['TotalPayBenefits'] = pd.to_numeric(sal['TotalPayBenefits'], errors='coerce')

# Filtrer SEULEMENT les lignes où TotalPayBenefits est un nombre POSITIF
# (C'est la clé ! Beaucoup de "Chief" ont TotalPayBenefits = NaN ou négatif)
sal_filtered = sal[sal['TotalPayBenefits'] > 0]

# Compter les "Chief" (insensible à la casse) dans ce sous-ensemble
chief_count = sal_filtered['JobTitle'].str.contains('Chief', case=False, na=False).sum()

print("Nombre de 'Chief' :", chief_count)
```

**Bonus : Y a-t-il une corrélation entre la longueur du titre de poste et le salaire ?**


```python
# Créer une colonne avec la longueur du titre de poste
sal['JobTitleLength'] = sal['JobTitle'].str.len()

# Calculer la corrélation entre la longueur du titre et le salaire total
correlation = sal['JobTitleLength'].corr(sal['TotalPayBenefits'])
print("Corrélation entre la longueur du titre et le salaire :", correlation)

# Interprétation simple :
if abs(correlation) < 0.1:
    print("→ Pas de corrélation significative.")
elif abs(correlation) < 0.3:
    print("→ Faible corrélation.")
else:
    print("→ Corrélation modérée à forte.")
```

```python
# Créer une nouvelle colonne avec la longueur du titre
sal['JobTitleLength'] = sal['JobTitle'].str.len()

# Sélectionner uniquement les deux colonnes d'intérêt
correlation_matrix = sal[['JobTitleLength', 'TotalPayBenefits']].corr()

# Afficher la matrice complète (comme dans l'image)
print("Matrice de corrélation :")
print(correlation_matrix)
```
