**Importer pandas et lire le fichier CSV des achats en ligne, puis le stocker dans un DataFrame nommé `ecom`.**


```python
import pandas as pd
```

```python
ecom = pd.read_csv("Ecommerce Purchases")
```

**Vérifier les premières lignes du DataFrame.**


```python
ecom.head()
```

**Combien y a-t-il de lignes et de colonnes ?**


```python
ecom.shape
```

**Quel est le prix moyen d’achat (`Purchase Price`) ?**


```python
float(ecom['Purchase Price'].mean())
```

**Quels étaient les prix d’achat les plus élevés et les plus bas ?**


```python
ecom['Purchase Price'].max()
```

```python
ecom['Purchase Price'].min()
```

**Combien de personnes ont choisi l’anglais ('en') comme langue sur le site web ?**


```python
ecom[ecom['Language'] == 'en'].shape[0]
```

**Combien de personnes ont le titre de poste "Lawyer" ?**


```python
ecom[ecom['Job'] == 'Lawyer'].shape[0]
```

**Combien de personnes ont effectué un achat le matin (AM) et combien l’ont fait l’après-midi (PM) ?**

*(Indice : consultez [value_counts()](http://pandas.pydata.org/pandas-docs/stable/generated/pandas.Series.value_counts.html)) *


```python
ecom['AM or PM'].value_counts()
```

**Quels sont les 5 titres de poste les plus courants ?**


```python
ecom['Job'].value_counts().head(5)
```

**Quel était le prix d’achat pour la transaction provenant du Lot : "90 WT" ?**


```python
float(ecom[ecom['Lot'] == '90 WT']['Purchase Price'].iloc[0])
```

**Quel est l’email de la personne ayant le numéro de carte de crédit suivant : 4926535242672853 ?**


```python
ecom[ecom['Credit Card'] == 4926535242672853]['Email'].iloc[0]
```

**Combien de personnes ont American Express comme fournisseur de carte de crédit *et* ont effectué un achat supérieur à 95 $ ?**


```python
ecom[(ecom['CC Provider'] == 'American Express') & (ecom['Purchase Price'] > 95)].shape[0]
```

**Difficile : Combien de personnes ont une carte de crédit qui expire en 2025 ?**


```python
ecom[ecom['CC Exp Date'].str.contains('/25')].shape[0]
```

**Difficile : Quels sont les 5 fournisseurs d’emails les plus populaires (par exemple gmail.com, yahoo.com, etc.) ?**


```python
ecom['Email'].str.split('@').str[1].value_counts().head(5)
```
