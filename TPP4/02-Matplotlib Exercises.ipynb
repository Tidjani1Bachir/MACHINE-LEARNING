```python
import matplotlib.pyplot as plt
import numpy as np

# Données
x = np.arange(0, 100)
y = x**2
z = x**3

# Créer la figure
fig = plt.figure()

# Ajouter le premier axe ax1 à la position [0, 0, 1, 1]
ax1 = fig.add_axes([0, 0, 1, 1])

# Ajouter le deuxième axe ax2 à la position [0.2, 0.5, 0.4, 0.4]
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])

# Tracer (x, y) sur les deux axes
ax1.plot(x, y)
ax2.plot(x, y)

# Définir les limites du zoom pour ax2 (comme visible dans l'image)
ax2.set_xlim(20, 22)
ax2.set_ylim(30, 50)

# Ajouter les labels et titres pour ax2 (zoom)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_title('zoom')

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

# Données
x = np.arange(0, 100)
y = x**2
z = x**3

# Créer la figure
fig = plt.figure()

# Ajouter le premier axe ax1 à la position [0, 0, 1, 1]
ax1 = fig.add_axes([0, 0, 1, 1])

# Ajouter le deuxième axe ax2 à la position [0.2, 0.5, 0.4, 0.4]
ax2 = fig.add_axes([0.2, 0.5, 0.4, 0.4])

# Tracer (x, z) sur le grand axe (ax1) - la courbe principale
ax1.plot(x, z, color='blue', linestyle='-')

# Tracer (x, y) sur le petit axe (ax2) - le zoom
ax2.plot(x, y, color='blue', linestyle='-')

# Définir les limites du zoom pour ax2 (comme visible dans l'image)
# L'image montre un zoom sur x entre 20 et 22, et y entre 30 et 50.
ax2.set_xlim(20, 22)
ax2.set_ylim(30, 50)

# Ajouter les labels et titres pour ax2 (zoom)
ax2.set_xlabel('x')
ax2.set_ylabel('y')
ax2.set_title('zoom')

plt.show()
```

** Now use x,y, and z arrays to recreate the plot below. Notice the xlimits and y limits on the inserted plot:**

```python
import matplotlib.pyplot as plt
import numpy as np

# Données
x = np.arange(0, 100)
y = x**2
z = x**3

# Créer une figure avec 1 ligne et 2 colonnes de sous-graphiques
fig, axes = plt.subplots(nrows=1, ncols=2)

# Tracer sur le premier sous-graphique (axes[0])
axes[0].plot(x, y)

# Tracer sur le deuxième sous-graphique (axes[1])
axes[1].plot(x, z)

plt.show()
```

## Exercise 4

** Use plt.subplots(nrows=1, ncols=2) to create the plot below.**

```python
import matplotlib.pyplot as plt
import numpy as np

# Données
x = np.arange(0, 100)
y = x**2
z = x**3

# Créer une figure avec 1 ligne et 2 colonnes de sous-graphiques
fig, axes = plt.subplots(nrows=1, ncols=2)

# Tracer (x, y) sur le premier sous-graphique avec une ligne bleue pointillée
axes[0].plot(x, y, color='blue', linestyle='--', linewidth=2)

# Tracer (x, z) sur le deuxième sous-graphique avec une ligne rouge pleine
axes[1].plot(x, z, color='red', linestyle='-', linewidth=2)

plt.show()
```

** Now plot (x,y) and (x,z) on the axes. Play around with the linewidth and style**

```python
import matplotlib.pyplot as plt
import numpy as np

# Données
x = np.arange(0, 100)
y = x**2
z = x**3

# Créer une figure avec 1 ligne et 2 colonnes, en spécifiant la taille (largeur, hauteur)
# Ici, figsize=(12, 5) donne une figure plus large, comme dans l'image.
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(12, 5))

# Tracer (x, y) sur le premier sous-graphique avec une ligne bleue pleine
axes[0].plot(x, y, color='blue', linestyle='-', linewidth=2)

# Tracer (x, z) sur le deuxième sous-graphique avec une ligne rouge pointillée
axes[1].plot(x, z, color='red', linestyle='--', linewidth=2)

plt.show()
```
