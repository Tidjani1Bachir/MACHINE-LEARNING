```python
# CODE ICI
# RECRÉEZ LE GRAPHIQUE DE L'EXERCICE CI-DESSOUS
# ATTENTION À NE PAS ÉCRASER LA CELLULE CI-DESSOUS
# SINON L'IMAGE DU GRAPHIQUE DE L'EXERCICE SERA PERDUE !
# CODE ICI
# RECRÉEZ LE GRAPHIQUE DE L'EXERCICE CI-DESSOUS
# ATTENTION À NE PAS ÉCRASER LA CELLULE CI-DESSOUS
# SINON L'IMAGE DU GRAPHIQUE DE L'EXERCICE SERA PERDUE !
sns.heatmap(titanic.select_dtypes(include='number').corr(), annot=True, cmap='coolwarm')
```

```python
# CODE ICI
# RECRÉEZ LE GRAPHIQUE DE L'EXERCICE CI-DESSOUS
# ATTENTION À NE PAS ÉCRASER LA CELLULE CI-DESSOUS
# SINON L'IMAGE DU GRAPHIQUE DE L'EXERCICE SERA PERDUE !
g = sns.FacetGrid(titanic, col="sex")
g.map(sns.histplot, "age")
```
